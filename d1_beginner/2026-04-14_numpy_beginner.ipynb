```python
# Drill 3
# array[start:stop:step]
# multi-dimensional array[row_slice, column_slice]
# note: numpy slicing ends are EXCLUSIVE --> [:2, :3] --> rows 0 and 1 with columns 0,1,and 2

import numpy as np

# 4x7 demand matrix
demand_matrix = np.array([
    [120, 85, 200, 150, 95, 180, 210],   # SKU-001
    [45,  60, 55,  70,  65, 50,  80],    # SKU-002
    [300, 280, 320, 290, 310, 295, 305], # SKU-003
    [15,  20,  18,  22,  19, 25,  21]    # SKU-004
])

# shape of the demand matrix
print(f"Shape: {demand_matrix}")
print(f"Rows (SKU's): {demand_matrix.shape[0]}")
print(f"Columns (days): {demand_matrix.shape[1]}")
print(f"\nFull matrix:\n{demand_matrix}")

# What does [1, :] mean?
# means select the first row and all columns
print(f"\nSKU-002 daily demand [1,:]: {demand_matrix[1, :]}")

# What does [:, 0] mean?
# means select all rows, and the first column since python is zero-based
print(f"\nAll SKUs day 1 [:, 0]: {demand_matrix[:, 0]}")

# What does [:2, :3] mean?
# means select all rows from 0 to 2 and select all columns from 0 to 3
print(f"\nFirst 2 SKU's, first 3 days [:2, :3]:\n {demand_matrix[:2, :3]}")
```

## Drill 5 - Boolean Indexing
- pandas uses `df[df["col"] > value]`
- numpy uses a similar pattern

```python
import numpy as np
demand = np.array([120, 45, 300, 0, 88, 210, 55, 175, 400, 30])
unit_cost = np.array([25.50, 42.00, 15.75, 15.75, 8.25, 8.25,
                      32.00, 18.50, 5.10, 5.10])

# which sku's have demand > 100
# create a mask
high_demand_mask = demand > 100
print(f"Boolean mask: {high_demand_mask}")

# apply the mask - returning only high demand values
high_demand = demand[high_demand_mask]
print(f"High demand SKU's: {high_demand}")

# complete the previous steps in a single-line
print(f"In one line:{demand[demand > 100]}")

# write a compound condition
affordable_high_dem = demand[(demand > 50) & (unit_cost < 20)]
print(f"High demand, low cost: {affordable_high_dem}")

# use np.where() to return index positions noting where the condition is true
high_demand_idx = np.where(demand > 100)
print(f"Indices where demand > 100: {high_demand_idx}")

# use np.where with conditions, critical if demand = 0, "ok" otherwise
status = np.where(demand == 0, "critical", "ok")
print(f"Stock status: {status}")
```

```python
# Drill 6: Array Reshaping
import numpy as np
# monthly demand values -- (12 rows, 0 columns)
monthly = np.array([120, 85, 200, 150, 95, 180,
                    210, 165, 245, 130, 175, 220])

# get the shape of the data (row,col)
print(f"Original shape: {monthly.shape}")
print(f"Original:\n{monthly}")

# reshape from (12,0) to (3,4)
quarterly = monthly.reshape(3,4)
print(f"\nReshaped to 3x4 (quarters x months):\n{quarterly}")

# reshape from (3,4) to (4,3)
by_quarter = monthly.reshape(4,3)
print(f"\nReshaped to 4 x 3:\n{by_quarter}")

# flatten back to one-dimension
flat = quarterly.flatten()
print(f"\nFlattened back: {flat}")

# use numpy to calculate the missing dimension
# the "-1" tells NumPy to figure out this dimension automatically based on the total number of elements
# and the other dimensions specified
auto = monthly.reshape(2, -1)
print(f"\nReshaped with -1 (2, auto):\n {auto}")
print(f"\nShape of auto:\n {auto.shape}")
```

```python
# Drill 7: Axis Aggregation

import numpy as np

# create the array of data
# 4SKU's x 7 days

demand_matrix = np.array([
    [120, 85, 200, 150, 95, 180, 210],   # SKU-001
    [45,  60, 55,  70,  65, 50,  80],    # SKU-002
    [300, 280, 320, 290, 310, 295, 305], # SKU-003
    [15,  20,  18,  22,  19, 25,  21]    # SKU-004
])

# axis=0 aggregates DOWN the columns
daily_total = demand_matrix.sum(axis=0)
print(f"Total demand per day (axis=0): {daily_total}")

# axis=1 aggregates across the rows
sku_total = demand_matrix.sum(axis=1)
print(f"Total demand per SKU: {sku_total}")

# find the mean per sku
sku_mean = demand_matrix.mean(axis=1).round(1)
print(f"Average daily demand per SKU: {sku_mean}")

# max demand for each sku
sku_peak = demand_matrix.max(axis=1)
print(f"Peak demand per SKU: {sku_peak}")

# which day had the highest demand for each SKU?
peak_day = demand_matrix.argmax(axis=1)
print(f"Peak demand day index per SKU: {peak_day}")

# normalize each SKU's daily demand as a % of its own total
# keepdims=True preserves the original number of dimensions after aggregating so (4, ) becomes (4,1)
normalized = (demand_matrix / demand_matrix.sum(axis=1, keepdims=True))
print(f"\nNormalized demand (% of SKU total):\n{normalized}")
```

```python
# Drill 8: Safety Stock Formula using NumPy
# Formula: Safety Stock = ZScore * SD_Demand * SQRT(Lead_Time)

# create the datasets
# SC dataset — 8 SKUs
sku_ids = np.array(["SKU-001", "SKU-002", "SKU-003", "SKU-004",
                    "SKU-005", "SKU-006", "SKU-007", "SKU-008"])

# service level z-scores
Z_SCORES = {0.90: 1.28, 0.95: 1.65, 0.99: 2.33}

daily_demand = np.array([120.0, 45.0, 300.0, 0.0, 88.0, 210.0, 55.0, 175.0])
demand_std   = np.array([15.2, 8.1, 42.3, 0.0, 12.5, 28.7, 9.3, 22.1])
lead_time    = np.array([14, 7, 21, 10, 5, 30, 12, 18])

# vectorize the function
# note: safety stock needs to always round up, for example 92.3 units, should round up to 93 to avoid shortages
def calculate_safety_stock(demand_std, lead_time, service_level=0.95):
    z = Z_SCORES[service_level]
    return np.ceil(z * demand_std * np.sqrt(lead_time)).astype(int)

# calculate the 95% service level
ss_95 = calculate_safety_stock(demand_std, lead_time, service_level=0.95)
print(f"Safety stock (95% Service Level):\n{ss_95}")

# calculate for all three service levels
for level, z in Z_SCORES.items():
    ss = calculate_safety_stock(demand_std, lead_time, service_level=level)
    print(f"\nService level {level*100: .0f}%:")
    for sku, stock in zip(sku_ids, ss):
        print(f"  {sku}: {stock} units")

# reorder point = safety stock + average demand during lead time
avg_demand_during_lt = daily_demand * lead_time
reorder_point = ss_95 + avg_demand_during_lt
print(f"\nReorder points (95% service level):\n {reorder_point.astype(int)}")
```
