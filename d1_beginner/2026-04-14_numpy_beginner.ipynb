```python
# Drill 1: Creating Arrays
import numpy as np

# create a simple 1-Dimensional Array - showing daily demand for a single SKU
demand = np.array([120, 85, 200, 150, 95, 180, 210])

print(f"Array: {demand}")
print(f"Type: {type(demand)}")
print(f"dtype: {demand.dtype}")
print(f"Shape: {demand.shape}")
print(f"Dimenions: {demand.ndim}")
print(f"Size: {demand.size}")
```

```python
# Drill 2: Array Creation Shortcuts

# Zeros - Initialize an empty inventory array
inventory = np.zeros(7)
print(f"Zeros: {inventory}")

# Ones - Initialize an array with all suppliers at 1
scores = np.ones(5)
print(f"Ones: {scores}")

# Generate a sequence -- np.arange() is inclusive and then exclusive
days = np.arange(start=1, stop=31)
print(f"Days: {days}")

# Generate evenly spaced values between two points
service_levels = np.linspace(start=0.90, stop=0.99, num=5)
print(f"Service levels: {service_levels}")

# Generate random integers meant to simulate lead times
lead_times = np.random.randint(low=3, high=30, size=10)
print(f"Lead times: {lead_times}")
```
